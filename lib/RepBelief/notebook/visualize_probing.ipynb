```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('seaborn')
plt.rcParams.update({'font.size': 22})
```

```python
def load_probing_results(dynamic='0_forward', variable='belief', belief='protagonist'):
    """프로빙 결과 파일들을 로드합니다."""
    base_path = './RepBelief/data/results/probe'
    
    # 각각의 메트릭에 대한 결과 로드
    val_acc = np.load(f'{base_path}/{dynamic}_{variable}_{belief}_val_acc.npy')
    coef = np.load(f'{base_path}/{dynamic}_{variable}_{belief}_coef.npy')
    
    return val_acc, coef
```

```python
def plot_probing_heatmap(data, title, save_path=None, vmin=0.5, vmax=1.0):
    """프로빙 결과를 히트맵으로 시각화합니다."""
    # 큰 사이즈의 figure 생성
    fig, ax = plt.subplots(figsize=(12, 10))
    
    # 히트맵 생성
    sns.heatmap(data, 
                ax=ax,
                cmap='Greens',
                vmin=vmin,
                vmax=vmax,
                cbar_kws={'drawedges': False},
                square=True)
    
    # 컬러바 커스터마이징
    cbar = ax.collections[0].colorbar
    cbar.outline.set_linewidth(2)
    
    # x축과 y축 틱 설정
    ax.set_xticks(np.arange(0.5, data.shape[1], 5))
    ax.set_yticks(np.arange(0.5, data.shape[0], 5))
    ax.set_xticklabels(np.arange(0, data.shape[1], 5), rotation=0)
    ax.set_yticklabels(np.arange(0, data.shape[0], 5))
    
    # 축 레이블 설정
    ax.set_xlabel('Head', fontsize=24, labelpad=20)
    ax.set_ylabel('Layer', fontsize=24, labelpad=20)
    
    # 타이틀 설정
    plt.title(title, fontsize=28, pad=20)
    
    # 테두리 설정
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(2)
    
    # 결과 저장
    if save_path:
        plt.savefig(save_path, format='pdf', bbox_inches='tight', dpi=300)
        
    plt.show()
```

```python
# 프로빙 결과 로드
dynamic = '0_forward'
variable = 'belief'
belief = 'protagonist'
```

```python
val_acc, coef = load_probing_results(dynamic, variable, belief)
```
