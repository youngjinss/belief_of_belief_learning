```python
import os
import sys
import time
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from matplotlib.lines import Line2D

# ray 관련 라이브러리 임포트
import ray
from ray import tune
from ray.tune.registry import register_env

# 프로젝트 루트 디렉토리 설정
os.chdir("/home/youngjins/project/belief_trading")

# lib 하위 경로 추가
sys.path.append("/home/youngjins/project/belief_trading/lib/")
sys.path.append("/home/youngjins/project/belief_trading/lib/abides_jpmc_public")

# 환경 임포트
import gym
from abides_gym.envs.markets_execution_environment_v1 import (
    SubGymMarketsExecutionEnv_v1,
)

# lib 모듈 임포트
from lib.utils import gym_types, flatten_dict, history_dict
from lib.policy import policyMACD
```

```python
# 매개 변수 선언
seed = 0
np.random.seed(seed)

# window size 선언
long_window = 20
short_window = 5
signal_window = 2
market_data_length = 10
parent_order_size = 2000
order_fixed_size = 10
mkt_close = "12:00:00",  # 왠진 모르겠는데, 10:00:00으로 하면, mkt_open이랑 차이가 별로 안남 --> mkt_open이 어디서 정의되는지 찾기 힘듬
execution_window = "12:00:00"

# 환경 선언
gym_type_str = "markets-execution-v1"
gym_type_abb = gym_types[gym_type_str]
register_env(
    gym_type_str,
    lambda config: SubGymMarketsExecutionEnv_v1(**config),
)

# ray init
ray.shutdown()
ray.init()
```

```python
# 강화학습 관련 변수 선언
env = gym.make(
    gym_type_str,
    background_config="rmsc04",
    timestep_duration="10S",
    execution_window=execution_window,
    mkt_close=mkt_close,
    parent_order_size=parent_order_size,
    order_fixed_size=order_fixed_size,
    not_enough_reward_update=-100,  # penalty
    state_history_length=long_window,
    market_data_buffer_length=market_data_length,
    debug_mode=True,
)
env.seed(seed)

target_agent = policyMACD(
    long_window=long_window, short_window=short_window, signal_window=signal_window, delta=0.05
)
```

```python
# 시뮬레이션 시작
state = env.reset()
done = False
episode_reward = 0

# 1개 에피소드 실행 (소요 시간 측정)
start_time = time.time()
while not done:
    action_ = target_agent.get_action(state)
    env.direction = (
        "BUY" if action_ > 0 else "SELL" if action_ < 0 else env.direction
    )  # 매수 / 매도 방향 설정

    # 방어 코드 시작
    # 매수 / 매도 크기 설정 (holdings = state[0] (holdings_pct) * parent_order_size)
    holdings = round(state[0][0] * parent_order_size)
    if action_ != 0 and holdings > 0:
        env.order_fixed_size = max(
            0, (int(holdings * abs(target_agent.trade_proportion) - holdings))
        )  # 매수 / 매도 크기 설정
    else:
        env.order_fixed_size = order_fixed_size

    # 매도 & env.order_fixed_size > holdings 인 경우, 매수 / 매도 크기를 holdings 만큼 줄임
    if action_ < 0 and holdings > 0 and env.order_fixed_size > holdings:
        env.order_fixed_size = holdings - 1

    # 매수 / 매도 크기가 0인 경우, action을 0으로 설정
    if env.order_fixed_size == 0:
        action_ = 0
        
    # holdings > parent_order_size 인 경우, action을 0으로 설정
    holdings = round(state[0][0] * parent_order_size)
    if action_ > 0 and holdings + env.order_fixed_size > parent_order_size:
        action_ = 0
    elif action_ < 0 and holdings - env.order_fixed_size < 0:
        action_ = 0
    # 방어 코드 종료

    action = abs(action_)
    state, reward, done, info = env.step(action)
    episode_reward += reward

    # 현재 상태에서 필요한 정보 추출
    history_dict["mid_price"].append((info["best_bid"] + info["best_ask"]) / 2)
    history_dict["best_bid"].append(info["best_bid"])
    history_dict["best_ask"].append(info["best_ask"])
    history_dict["reward"].append(reward)
    history_dict["cumulative_reward"].append(episode_reward)
    history_dict["pnl"].append(info["pnl"])
    history_dict["holdings"].append(info["holdings"])
    history_dict["time"].append(info["current_time"])
    history_dict["action"].append(action_)
    history_dict["order_size"].append(env.order_fixed_size)  # debuging
    history_dict["direction"].append(env.direction)  # debuging

end_time = time.time()
print(f"Execution time: {end_time - start_time} seconds")

# could add a few more...
output = flatten_dict(info)
output["episode_reward"] = episode_reward
output["name"] = target_agent.name
```

```python
# 그래프 설정
fig, axes = plt.subplots(5, 1, figsize=(10, 10), sharex=True)

# 폰트 사이즈 조정
font_size = 12
axes[0].tick_params(labelsize=font_size)
axes[1].tick_params(labelsize=font_size)
axes[2].tick_params(labelsize=font_size)
axes[3].tick_params(labelsize=font_size)
axes[4].tick_params(labelsize=font_size)

# 범례 폰트 사이즈 조정
legend_font_size = 10

# 특정 기간 설정
start_period = 0  # 시작 인덱스
end_period = 100000  # len(history_dict["best_bid"])  # 끝 인덱스 (기본값은 전체 데이터)

# 1. 가격 변동 그래프
axes[0].plot(range(start_period, min(end_period, len(history_dict["best_bid"]))), 
             history_dict["mid_price"][start_period:end_period], label="Mid Price", color="black")
axes[0].scatter(
    range(start_period, min(end_period, len(history_dict["best_bid"]))),
    history_dict["best_bid"][start_period:end_period],
    label="Best Bid",
    alpha=0.7,
)
axes[0].scatter(
    range(start_period, min(end_period, len(history_dict["best_bid"]))),
    history_dict["best_ask"][start_period:end_period],
    label="Best Ask",
    alpha=0.7,
)
axes[0].set_title("ABIDES Simulation")
axes[0].set_ylabel("Price")
axes[0].legend()
axes[0].grid(True)

# 2. 보유량 그래프
axes[1].plot(range(start_period, min(end_period, len(history_dict["best_bid"]))),
             history_dict["holdings"][start_period:end_period], label="Holdings")
axes[1].set_title("Stock Holding Variation")
axes[1].set_ylabel("Quantity")
axes[1].legend()
axes[1].grid(True)

# 3. 수익 그래프
# 왼쪽 y축은 Cumulative Reward용
axes[2].plot(range(start_period, min(end_period, len(history_dict["best_bid"]))),
            np.array(history_dict["pnl"][start_period:end_period]).cumsum(),
            label="Cumulative Return",
            color="green",
)
axes[2].set_title("Profit and Loss (PnL)")
axes[2].set_ylabel("Cumulative Return")
axes[2].tick_params(axis="y", labelcolor="green")
axes[2].grid(True)

# 오른쪽 y축은 PnL용
ax2 = axes[2].twinx()
ax2.bar(range(start_period, min(end_period, len(history_dict["best_bid"]))), 
        history_dict["pnl"][start_period:end_period], 
        label="PnL", color="purple", alpha=0.5)
ax2.set_ylabel("Return")
ax2.tick_params(axis="y", labelcolor="purple")

# 두 축의 범례를 하나로 합치기
lines1, labels1 = axes[2].get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
axes[2].legend(lines1 + lines2, labels1 + labels2, loc="best")

# 4. 행동 그래프 (action > 0 이면 빨간색 위 방향 삼각형, action < 0 이면 파란색 아래 방향 삼각형)
# 행동 그래프를 첫 번째 가격 그래프에 추가
for i in range(start_period, min(end_period, len(history_dict["best_bid"]))):
    action = history_dict["action"][i]
    if action > 0:  # 매수 행동
        axes[3].scatter(
            i, history_dict["mid_price"][i], marker="^", color="red", s=100, alpha=0.7
        )
    elif action < 0:  # 매도 행동
        axes[3].scatter(
            i, history_dict["mid_price"][i], marker="v", color="blue", s=100, alpha=0.7
        )

# 범례에 행동 표시 추가
custom_lines = [
    Line2D([0], [0], marker="^", color="red", linestyle="None", markersize=8),
    Line2D([0], [0], marker="v", color="blue", linestyle="None", markersize=8),
]
axes[3].plot(range(start_period, min(end_period, len(history_dict["best_bid"]))), 
             history_dict["mid_price"][start_period:end_period], label="Mid Price", color="black")
long_ema = pd.Series(history_dict["mid_price"]).ewm(span=long_window, adjust=False).mean()
short_ema = pd.Series(history_dict["mid_price"]).ewm(span=short_window, adjust=False).mean()
axes[3].plot(range(start_period, min(end_period, len(history_dict["best_bid"]))), 
             long_ema[start_period:end_period], label="Long EMA", color="orange", alpha=0.5)
axes[3].plot(range(start_period, min(end_period, len(history_dict["best_bid"]))), 
             short_ema[start_period:end_period], label="Short EMA", color="blue", alpha=0.5)
axes[3].legend(handles=custom_lines, labels=["Buy", "Sell"], fontsize=legend_font_size)
axes[3].set_xlabel("Time")
axes[3].grid(True)
axes[3].set_ylabel("Positions")

# MACD 그래프 추가 - mid_price 에서 ema with long_window - ema with short_window 의 차이
macd_line = (long_ema - short_ema).ewm(span=signal_window, adjust=False).mean()
signal = macd_line[1:].reset_index(drop=True) - macd_line[:-1].reset_index(drop=True)
axes[4].plot(range(start_period + 1, min(end_period, len(history_dict["best_bid"]))), 
             signal[start_period:end_period], label="MACD")
axes[4].set_title("MACD")
axes[4].set_ylabel("MACD")
axes[4].legend()
axes[4].grid(True)

plt.tight_layout()
plt.show()
```
