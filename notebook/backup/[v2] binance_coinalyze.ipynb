```python
import pandas as pd
import requests
import datetime


def get_ratio(symbol, interval, start_time, end_time):
    url = "https://api.coinalyze.net/v1/long-short-ratio-history"
    headers = {"api_key": "a434a452-56f3-47c6-aa64-1fa681cd336b"}
    params = {
        "symbols": symbol,
        "interval": interval,
        "from": start_time,
        "to": end_time,
    }
    res = requests.get(url, headers=headers, params=params)

    return res.json()
```
