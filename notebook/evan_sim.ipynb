```python
import sys
import os

# 현재 노트북 디렉토리의 상위 디렉토리를 시스템 경로에 추가
sys.path.append(os.path.abspath('..'))

import matplotlib.pyplot as plt

# script 디렉토리에서 simulation 모듈 가져오기
from lib.backup.evan_2024 import create_simple_market

# 간단한 시장 시뮬레이션 생성 및 실행
print("간단한 시장 시뮬레이션 실행 중...")
market = create_simple_market()
market.run_simulation(3000)

# 포트폴리오 수렴 분석
portfolios = market.analyze_portfolio_convergence()

# 포트폴리오 특성 시각화
plt.figure(figsize=(10, 6))
for p in portfolios:
    plt.scatter(p['variance'], p['expected_return'], alpha=0.5)
plt.title('포트폴리오 평균-분산 특성')
plt.xlabel('포트폴리오 분산')
plt.ylabel('예상 포트폴리오 수익률')
plt.show()
```
